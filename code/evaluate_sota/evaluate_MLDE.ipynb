# ftMLDE
## Notes:
MSA Transformer encoding works the best according to the paper.
-> use argument `encoding transformer`

## Steps:
### Build alignment for MSA transfromer
"If using the MSA Transformer from ESM (esm_msa1_t12_100M_UR50S), then this should be a .a2m or .a3m file where the first sequence in the alignment is the reference sequence. See the section on Building an Alignment for MSA Transformer for instructions on preparing an msa for an input."

### Generate encoding
`generate_encoding.py --encoding transformer --protein_name GB1 --fasta <path_to_wt_sequence.fasta> --positions Q3 F97 --n_combined 2`
- the fasta file only contains the wt sequece

### Get params for EVmutation model
see the original paper: `https://www.nature.com/articles/nbt.3769`
input for the parameter-building is the sequence alignment

### Zero shot prediction
`predict_zero_shot.py --positions Q3 F97 --models ?esm_msa1_t12_100M_UR50S --fasta <path_to_wt_sequence.fasta> --alignment <path_to_alignment.a3m> --evmutation_model ??`
- the alignment file has to be a2m or a3m, first sequence is the wt sequence
- `esm_msa1_t12_100M_UR50S` probably is what AFP-DE paper means by "trans." sampling

- could also try using deep_sequence ?

### Training & mlde
`execute_mdle.py --training_data ?? --combo_to_ind_dict ??`

- main output: `PredictedFitness.csv`
- take top n variants with highest predicted fitness and evaluate their real fitness
