BO = "NaiveBO" in the paper - classical BO
TuRBO = Trust Region BO
ODBO = combines BO with outlier detection (XGBOD)

gp_regression = classical gp regression
robust_regression = RobustGP

Notable differences in our way of running ODBO

1)
They had a save preselected 40 sequences. (suspicious)
We use their sampling to sample different 40 sequences in each of the 100 runs.

2)
They only let the algorithm choose from the experimentally measured 149361 variants.
We let the algorithm choose from all 160000 variants (more realistic use of the algorithm).
The unmeasured variants are considered to have 0.0 fitness. (Done in previous studies and has some biological motivation.)
