```python
import pandas as pd
```

```python
dataset_name = "TrpB" # GB1, PhoQ, TrpB
```

```python
# Load the CSV file
file_path = "../../data/"+dataset_name+"/esm-1b_variants_complete.csv"
variants = pd.read_csv(file_path)
```

```python
ALPHABET = ['A', 'R', 'N', 'D', 'C', 'Q', 'E', 'G', 'H', 'I', 'L', 'K', 'M', 'F', 'P', 'S', 'T', 'W', 'Y', 'V']

import numpy as np

# Create a mapping from each character in ALPHABET to an index
char_to_index = {char: idx for idx, char in enumerate(ALPHABET)}

# Function to one-hot encode a single sequence
def one_hot_encode(sequence, alphabet_size=len(ALPHABET)):
    encoding = np.zeros((len(sequence), alphabet_size), dtype=int)
    for i, char in enumerate(sequence):
        encoding[i, char_to_index[char]] = 1
    return encoding

variants = [[c for c in v][0] for v in variants.values]

one_hot_encoded_variants = [one_hot_encode(seq) for seq in variants]
```

```python
import csv

# Save one_hot_encoded_variants to a CSV file
output_file_path = "../../data/"+dataset_name+"/onehot_encoding_complete.csv"

with open(output_file_path, mode='w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow([f"Position_{i+1}" for i in range(len(ALPHABET) * len(variants[0]))])
    for encoding in one_hot_encoded_variants:
        # Flatten the 2D array into a 1D list for each sequence
        flattened_encoding = encoding.flatten()
        writer.writerow(flattened_encoding)
```
